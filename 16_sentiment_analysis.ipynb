````python
samples = ["I absolutely love this product!", "This is the worst experience.", "The product arrived today."]
try:
    from textblob import TextBlob
    for s in samples:
        p = TextBlob(s).sentiment.polarity
        print("TextBlob:", s, "->", p, "positive" if p>0 else "negative" if p<0 else "neutral")
except ImportError:
    print("Install textblob with pip install textblob")
try:
    from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
    analyzer = SentimentIntensityAnalyzer()
    for s in samples: print("VADER:", s, analyzer.polarity_scores(s))
except ImportError:
    print("VADER package missing. Install with: pip install vaderSentiment")
````
