Transformer-based pretrained models may require `pip install transformers torch sentence-transformers` and internet access to download weights.

````python
text = "Hello, how are you?"
try:
    from transformers import pipeline
    translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")
    print(translator(text, max_length=60))
except Exception as e:
    print("MarianMT not available offline:", type(e).__name__)
    # A tiny dictionary fallback makes the workflow executable without claiming full translation.
    dictionary = {"hello":"bonjour", "how":"comment", "are":"êtes", "you":"vous"}
    print("Dictionary demo:", " ".join(dictionary.get(w.lower().strip("?,"), w) for w in text.split()))
print("Actual MarianMT/T5 translation requires transformers and downloading pretrained weights.")
````
