````python
sentences = [["nlp","uses","machine","learning"], ["machine","learning","models"],
             ["word","embeddings","represent","words"], ["nlp","models","understand","text"]]
try:
    from gensim.models import Word2Vec
    model = Word2Vec(sentences, vector_size=30, window=3, min_count=1, workers=1, epochs=100, seed=42)
    print("Vector for 'machine' (first 8 values):", model.wv["machine"][:8].round(3))
    print("Similar words:", model.wv.most_similar("machine", topn=3))
except Exception as e:
    print("Gensim could not load in this environment:", repr(e))
    print("Install compatible versions with: pip install gensim")
````
