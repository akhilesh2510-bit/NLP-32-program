````python
text = "Google was founded in California. Sundar Pichai leads the company."
import spacy
nlp = spacy.blank("en")
doc = nlp(text)
print("spaCy blank pipeline entities:", [(e.text, e.label_) for e in doc.ents])
print("A trained NER model is required for predictions. Install with:")
print("python -m spacy download en_core_web_sm")
````
