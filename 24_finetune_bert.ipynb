````python
# Offline-executable baseline plus optional Hugging Face BERT fine-tuning recipe.
texts = ["This movie is excellent", "I loved the film", "This movie is terrible", "I hated the film",
         "What a wonderful story", "A boring and bad movie"]
labels = [1,1,0,0,1,0]
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
baseline = make_pipeline(TfidfVectorizer(), LogisticRegression(random_state=42))
baseline.fit(texts, labels)
print("Offline baseline prediction:", baseline.predict(["excellent and wonderful film"]).tolist())
try:
    import transformers, torch
    print("Transformers available:", transformers.__version__)
    print("For full BERT fine-tuning, use AutoTokenizer, AutoModelForSequenceClassification and Trainer with a larger labelled dataset.")
except ImportError:
    print("Hugging Face Transformers not installed. Optional: pip install transformers datasets accelerate torch")
````
