````python
text = "Natural Language Processing is fascinating. It helps computers understand human language!"
# NLTK (uses its built-in tokenizer; no downloaded corpus required)
try:
    from nltk.tokenize import sent_tokenize, word_tokenize
    try:
        sentences = sent_tokenize(text)
    except LookupError:
        from nltk.tokenize import PunktSentenceTokenizer
        sentences = PunktSentenceTokenizer().tokenize(text)
    try:
        words = word_tokenize(text)
    except LookupError:
        from nltk.tokenize import TreebankWordTokenizer
        words = TreebankWordTokenizer().tokenize(text)
    print("NLTK sentences:", sentences)
    print("NLTK words:", words)
except ImportError:
    print("Install NLTK with: pip install nltk")
# spaCy blank English pipeline works without downloading a language model
try:
    import spacy
    nlp = spacy.blank("en")
    doc = nlp(text)
    print("spaCy sentences:", [s.text for s in doc.sents] if doc.has_annotation("SENT_START") else "Use sentencizer")
    if not doc.has_annotation("SENT_START"):
        nlp.add_pipe("sentencizer")
        doc = nlp(text)
    print("spaCy sentences:", [s.text for s in doc.sents])
    print("spaCy words:", [t.text for t in doc])
except ImportError:
    print("Install spaCy with: pip install spacy")
````
