````python
# Small offline rule-based dialogue loop; optional DialoGPT code path shown below.
responses = {"hello":"Hello! How can I help you?", "name":"I am an NLP demo chatbot.",
             "bye":"Goodbye! Have a great day!"}
def chatbot(message):
    key = message.lower()
    for trigger, reply in responses.items():
        if trigger in key: return reply
    return "I am a demo bot. Ask me hello, name, or bye."
for q in ["hello", "what is your name?", "bye"]:
    print("User:", q, "\\nBot:", chatbot(q))
try:
    import transformers
    print("Transformers installed. DialoGPT can be loaded with AutoModelForCausalLM('microsoft/DialoGPT-small').")
except ImportError:
    print("Optional: pip install transformers torch; model weights require internet.")
````
