````python
# A compact NumPy recurrent sequence classifier demo (runs offline).
import numpy as np
sequences = np.array([[1,2,0],[2,3,0],[8,9,0],[9,8,0]], dtype=float)
labels = np.array([0,0,1,1])
Wx = np.array([[.2,-.1],[.1,.2],[0,0]])
Wh = np.eye(2)*.5
Wout = np.array([[1.0],[-1.0]])
for seq in sequences:
    h = np.zeros(2)
    for token in seq:
        x = np.array([token/10, 1.0, 0.0])
        h = np.tanh(x @ Wx + h @ Wh)
    score = float(h @ Wout)
    print("Sequence:", seq.tolist(), "score:", round(score,3), "class:", int(score<0))
print("This illustrates recurrent state updates. For a trainable LSTM, install tensorflow and define an Embedding-LSTM-Dense model.")
````
