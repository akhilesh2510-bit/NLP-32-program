````python
import re
sentence = "Riya bought 3 red apples from the market."
pattern = r"(?P<NUMBER>\\b\\d+\\b)|(?P<ADJECTIVE>\\b(red|green|fresh)\\b)|(?P<NOUN>\\b(apples|market)\\b)"
print("Regex chunks:")
for m in re.finditer(pattern, sentence, flags=re.I):
    print(m.lastgroup, ":", m.group())
# spaCy noun chunks require a trained parser; demonstrate tokenization with blank pipeline.
import spacy
nlp = spacy.blank("en")
doc = nlp(sentence)
print("spaCy tokens:", [(t.text, t.is_alpha) for t in doc])
print("Note: noun_chunks needs an English model with a dependency parser.")
````
