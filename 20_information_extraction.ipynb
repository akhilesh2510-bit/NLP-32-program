````python
import re
document = """Invoice No: INV-2041
Customer: Riya Sharma
Email: riya@example.com
Date: 04-11-2026
Total: INR 12500"""
patterns = {"invoice":r"Invoice No:\\s*(\\S+)", "customer":r"Customer:\\s*(.+)",
            "email":r"[\\w.-]+@[\\w.-]+", "date":r"\\d{2}-\\d{2}-\\d{4}", "total":r"Total:\\s*(.+)"}
for key, pat in patterns.items():
    m = re.search(pat, document)
    print(key, ":", m.group(1) if m and m.lastindex else m.group(0) if m else "Not found")
````
