````python
import numpy as np
query = np.array([1.0, 0.0])
keys = np.array([[1,0],[0,1],[1,1]], dtype=float)
values = np.array([[10,0],[0,10],[5,5]], dtype=float)
scores = keys @ query / np.sqrt(query.size)
weights = np.exp(scores-scores.max()); weights /= weights.sum()
context = weights @ values
print("Attention scores:", scores.round(3))
print("Attention weights:", weights.round(3))
print("Context vector:", context.round(3))
````
