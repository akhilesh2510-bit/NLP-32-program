````python
words = ["playing", "played", "studies", "studying", "better", "cars"]
from nltk.stem import PorterStemmer, WordNetLemmatizer
stemmer = PorterStemmer()
print("Stemming:")
for w in words: print(w, "->", stemmer.stem(w))
print("\\nLemmatization (WordNet corpus may be unavailable):")
try:
    from nltk.corpus import wordnet
    lemma = WordNetLemmatizer()
    for w in words: print(w, "->", lemma.lemmatize(w))
except LookupError:
    # Transparent fallback: spaCy rule-free token normalization is not a true lemmatizer.
    print("WordNet data not installed. Run nltk.download('wordnet') for full lemmatization.")
````
