````python
sentence = "The quick brown fox jumps over the lazy dog."
tokens = sentence.replace(".", " .").split()
# NLTK tagger needs downloaded models; use it when available, otherwise show a clear heuristic demo.
try:
    import nltk
    try:
        tagged = nltk.pos_tag(nltk.word_tokenize(sentence))
        print("NLTK POS tags:", tagged)
    except LookupError:
        print("NLTK POS model is not downloaded. Heuristic POS demo:")
        tagged = [(w, "PUNCT" if w == "." else "DET" if w.lower() in {"the","a","an"} else
                   "VERB" if w.lower() in {"jumps","runs","is"} else
                   "ADJ" if w.lower() in {"quick","brown","lazy"} else "NOUN") for w in tokens]
        print(tagged)
except ImportError:
    print("Install nltk to use nltk.pos_tag.")
````
