````python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
texts = ["win money now", "claim your free prize", "meeting at 10 am", "please send the report",
         "exclusive cash offer", "project meeting tomorrow", "free gift click here", "review the document"]
labels = ["spam","spam","ham","ham","spam","ham","spam","ham"]
Xtr, Xte, ytr, yte = train_test_split(texts, labels, test_size=.25, random_state=7, stratify=labels)
for name, clf in [("Naive Bayes", MultinomialNB()), ("SVM", LinearSVC())]:
    model = Pipeline([("tfidf", TfidfVectorizer()), ("clf", clf)])
    model.fit(Xtr,ytr)
    pred = model.predict(Xte)
    print(name, "accuracy:", round(accuracy_score(yte,pred),2))
    print(classification_report(yte,pred,zero_division=0))
````
