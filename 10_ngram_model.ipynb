````python
from sklearn.feature_extraction.text import CountVectorizer
corpus = ["natural language processing is useful", "language models process text"]
for n, label in [(1,"Unigrams"), (2,"Bigrams"), (3,"Trigrams")]:
    cv = CountVectorizer(ngram_range=(n,n), token_pattern=r"(?u)\b\w+\b")
    mat = cv.fit_transform(corpus)
    print(label, dict(zip(cv.get_feature_names_out(), mat.toarray().sum(axis=0))))
````
