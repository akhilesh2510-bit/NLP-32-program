````python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
docs = ["cats chase mice", "dogs chase balls", "cats and dogs are pets"]
bow = CountVectorizer()
tfidf = TfidfVectorizer()
print("BoW vocabulary:", bow.fit_transform(docs).toarray())
print("TF-IDF matrix:\\n", tfidf.fit_transform(docs).toarray().round(3))
print("BoW counts frequency; TF-IDF downweights terms common across documents.")
````
