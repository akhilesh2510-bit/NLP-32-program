````python
# Offline fallback: a tiny next-word Markov generator trained on a custom corpus.
corpus = ["natural language models generate text", "language models learn patterns", "models generate natural text"]
from collections import defaultdict, Counter
next_words = defaultdict(Counter)
for line in corpus:
    w = ["<START>"] + line.split() + ["<END>"]
    for a,b in zip(w,w[1:]): next_words[a][b] += 1
def generate(seed="natural", steps=8):
    out=[seed]; current=seed
    for _ in range(steps):
        choices=next_words.get(current)
        if not choices: break
        current=choices.most_common(1)[0][0]
        if current=="<END>": break
        out.append(current)
    return " ".join(out)
print("Generated demo:", generate())
print("Optional GPT-2: install transformers, load AutoModelForCausalLM('gpt2'); pretrained weights need internet.")
````
