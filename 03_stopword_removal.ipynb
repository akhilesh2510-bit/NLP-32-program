````python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
document = "This is a simple example showing how we can remove common words from a document."
tokens = [w.strip(".,!?").lower() for w in document.split()]
filtered = [w for w in tokens if w and w not in ENGLISH_STOP_WORDS]
print("Original:", tokens)
print("After stop-word removal:", filtered)
````
