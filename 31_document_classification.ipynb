````python
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
docs = ["The team won the football match", "Player scored a goal", "The stock market rose today",
        "Investors bought company shares", "Python code trains a machine learning model",
        "Neural networks learn patterns"]
labels = ["Sports","Sports","Finance","Finance","Technology","Technology"]
model = make_pipeline(TfidfVectorizer(), MultinomialNB()).fit(docs, labels)
new_docs = ["The striker scored two goals", "AI model uses Python", "Company shares increased"]
for d, c in zip(new_docs, model.predict(new_docs)):
    print(f"{c:12} <- {d}")
````
