````python
article = """Natural language processing is a field of artificial intelligence.
It helps computers understand, interpret and generate human language.
Applications include translation, sentiment analysis and question answering.
Modern transformer models have improved many NLP tasks."""
try:
    from transformers import pipeline
    summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")
    print(summarizer(article, max_length=45, min_length=15, do_sample=False))
except Exception as e:
    # Extractive offline summary: choose first two sentences.
    sentences = [s.strip() for s in article.split(".") if s.strip()]
    print("Extractive fallback summary:", ". ".join(sentences[:2]) + ".")
    print("For BART/T5 abstractive summarization, install transformers and download model weights.")
````
