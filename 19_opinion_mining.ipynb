````python
import pandas as pd
reviews = pd.DataFrame({"review":["Amazing battery and fast delivery","Poor quality, very disappointing",
                                  "Good value for money","Terrible customer support"],
                        "rating":[5,1,4,1]})
reviews["opinion"] = reviews.rating.map(lambda x: "positive" if x>=4 else "negative" if x<=2 else "neutral")
print(reviews)
print("\\nOpinion counts:\\n", reviews.opinion.value_counts())
````
