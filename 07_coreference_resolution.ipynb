````python
text = "Aarav bought a laptop. He uses it for machine learning."
# Lightweight demonstration of pronoun-to-entity linking; full neural coreference needs a pretrained model.
sentences = text.split(". ")
antecedents = {"he": "Aarav", "it": "a laptop"}
resolved = []
for sentence in sentences:
    words = sentence.split()
    resolved.append(" ".join(antecedents.get(w.lower().strip(".,!?"), w) for w in words))
print("Original:", text)
print("Simple resolved demo:", ". ".join(resolved))
print("For production coreference, use a neuralcoref-compatible spaCy model or a Hugging Face coreference model.")
````
