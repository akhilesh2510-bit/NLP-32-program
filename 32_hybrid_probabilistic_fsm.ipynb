````python
# Sentence-level boundary detection using keyword probabilities + a finite-state machine.
lines = ["INVOICE", "Invoice Number: A-102", "Date: 04-11-2026", "Total: 5000",
         "Thank you", "REPORT", "Quarterly sales increased", "Revenue: 90000"]
boundary_words = {"invoice", "report", "receipt", "statement"}
state = "OUTSIDE"
documents=[]
current=[]
for line in lines:
    tokens = set(line.lower().replace(":","").split())
    p_boundary = 0.9 if tokens & boundary_words else 0.1
    is_header = p_boundary >= 0.5
    if is_header and state == "INSIDE" and current:
        documents.append(current); current=[]
    if is_header: state = "INSIDE"
    if state == "INSIDE": current.append(line)
if current: documents.append(current)
for i, doc in enumerate(documents, 1):
    print(f"Document {i}:")
    print("\\n".join(doc))
print("FSM states: OUTSIDE -> INSIDE when boundary probability crosses threshold.")
````
