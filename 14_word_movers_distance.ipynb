````python
# WMD requires word embeddings. This small 2D example computes the optimal transport
# cost for a one-to-one toy alignment, illustrating the distance concept without external files.
import numpy as np
from scipy.spatial.distance import cdist
from scipy.optimize import linear_sum_assignment
vocab = {"cat": [0,0], "kitten": [0.2,0], "dog": [2,0], "puppy": [2.2,0]}
a, b = ["cat","dog"], ["kitten","puppy"]
cost = cdist(np.array([vocab[w] for w in a]), np.array([vocab[w] for w in b]))
r, c = linear_sum_assignment(cost)
print("Word distance matrix:\\n", cost.round(3))
print("Toy WMD alignment cost:", cost[r,c].sum())
print("For true WMD use normalized word frequencies and an Earth Mover's Distance solver.")
````
