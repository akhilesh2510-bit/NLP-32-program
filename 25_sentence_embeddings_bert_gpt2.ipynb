Transformer-based pretrained models may require `pip install transformers torch sentence-transformers` and internet access to download weights.

````python
sentences = ["Natural language processing is useful.", "NLP helps computers process language."]
try:
    from sentence_transformers import SentenceTransformer
    model = SentenceTransformer("all-MiniLM-L6-v2")
    embeddings = model.encode(sentences)
    print("Sentence embedding shape:", embeddings.shape)
except Exception as e:
    # Offline fallback: TF-IDF sentence vectors
    from sklearn.feature_extraction.text import TfidfVectorizer
    embeddings = TfidfVectorizer().fit_transform(sentences).toarray()
    print("SentenceTransformer unavailable; TF-IDF fallback shape:", embeddings.shape)
print("To generate actual BERT/GPT-2 representations install sentence-transformers and transformers; pretrained weights require internet on first run.")
````
