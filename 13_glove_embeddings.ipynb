````python
from pathlib import Path
glove_text = "king 0.8 0.2 0.1\nqueen 0.7 0.3 0.2\ncomputer 0.1 0.9 0.4\n"
Path("sample_glove.txt").write_text(glove_text)
embeddings = {}
with open("sample_glove.txt") as f:
    for line in f:
        parts = line.split()
        embeddings[parts[0]] = [float(x) for x in parts[1:]]
print("Loaded words:", list(embeddings))
print("Vector for king:", embeddings["king"])
print("Dimension:", len(embeddings["king"]))
````
