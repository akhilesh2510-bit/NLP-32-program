````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
docs = ["football team scored a goal", "players train for football", "AI learns from training data",
        "machine learning models use data", "goal scored by the team"]
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(docs)
lsa = TruncatedSVD(n_components=2, random_state=42).fit(X)
terms = tfidf.get_feature_names_out()
for i, comp in enumerate(lsa.components_):
    print("Topic", i+1, [terms[j] for j in comp.argsort()[-5:][::-1]])
print("Document-topic shape:", lsa.transform(X).shape)
````
