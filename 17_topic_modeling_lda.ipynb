````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
docs = ["team won the football match", "players scored goals in football", "stocks and market prices rose",
        "investors study stock market", "AI models learn from data", "machine learning uses data"]
cv = CountVectorizer(stop_words="english")
X = cv.fit_transform(docs)
lda = LatentDirichletAllocation(n_components=3, random_state=42, max_iter=20)
lda.fit(X)
terms = cv.get_feature_names_out()
for i, topic in enumerate(lda.components_):
    top = topic.argsort()[-5:][::-1]
    print("Topic", i+1, ":", [terms[j] for j in top])
````
