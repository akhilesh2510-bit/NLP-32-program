````python
text = "Sundar Pichai works at Google in California."
try:
    from transformers import pipeline
    ner = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
    print(ner(text))
except Exception as e:
    print("Transformer NER not run (package/model may be unavailable):", type(e).__name__)
    print("Offline demonstration: regex-based organization/location hints")
    import re
    print("Capitalized phrases:", re.findall(r"\\b[A-Z][a-z]+(?:\\s+[A-Z][a-z]+)*\\b", text))
print("Install transformers and allow model download for actual BERT NER.")
````
